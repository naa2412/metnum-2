## Gradient Descent

```python
import numpy as np
import pandas as pd

# data x1, x2, x3
X_raw = np.array([
    [30, 9, 90],
    [20, 9, 80],
    [25, 3, 90]
], dtype=float)

# data target y
y_raw = np.array([92, 70, 85], dtype=float)

# data baru yang ingin diprediksi
X_new_raw = np.array([
    [27, 6, 85]
], dtype=float)

# normalisasi menggunakan min-max scaling
def minmax_fit(data):
  data_min = data.min(axis=0)
  data_max = data.max(axis=0)
  return data_min, data_max

def minmax_transform(data, data_min, data_max):
  return (data-data_min)/(data_max - data_min)

def minmax_inverse(data_norm, data_min, data_max):
  return data_norm*(data_max - data_min) + data_min

# normalisasi x
X_min, X_max = minmax_fit(X_raw)
X = minmax_transform(X_raw, X_min, X_max)

# normalisasi y
y_min, y_max = minmax_fit(y_raw)
y = minmax_transform(y_raw, y_min, y_max)

# normalisasi data baru
X_new = minmax_transform(X_new_raw, X_min, X_max)

# model
# y_hat = w1*x1 + w2*x2 + w3*x3 + b
def predict(X, w, b):
  return X @ w + b

# loss function
# f(w, b) = sum((y_hat - y)^2)
def loss_function(X, y, w, b):
  y_pred = predict(X, w, b)
  error = y_pred - y
  return np.sum(error ** 2)

# gradien
def gradient(X, y, w, b):
  y_pred = predict(X, w, b)
  error = y_pred - y

  grad_w = 2 * X.T @ error
  grad_b = 2 * np.sum(error)

  return grad_w, grad_b

# gradient descent dengan bias
def gradient_descent(X, y, learning_rate=0.1, max_iter=10000, tolerance=1e-10):
  jumlah_fitur = X.shape[1]
  w = np.zeros(jumlah_fitur)
  b = 0.0

  loss_history = []

  for i in range(max_iter):
    loss = loss_function(X, y, w, b)
    loss_history.append(loss)

    grad_w, grad_b = gradient(X, y, w, b)

    w_new = w - learning_rate * grad_w
    b_new = b - learning_rate * grad_b

    if np.linalg.norm(w_new - w) < tolerance and abs(b_new - b) < tolerance:
      w = w_new
      b = b_new
      break

    w = w_new
    b = b_new

  return w, b, loss_history

# training model
w_gd, b_gd, loss_gd = gradient_descent(
    X, y,
    learning_rate=0.1,
    max_iter=10000,
    tolerance=1e-10
)

# prediksi
y_pred_gd_norm = predict(X_new, w_gd, b_gd)

# kembalikan ke skala persen
y_pred_gd = minmax_inverse(y_pred_gd_norm, y_min, y_max)

# output
print("=== DATA SETELAH NORMALISASI ===")

X_norm_df = pd.DataFrame(X, columns=["x1", "x2", "x3"])
y_norm_df = pd.DataFrame(y, columns=["y"])

print("\nX normalized:")
print(X_norm_df.to_string(index=False, float_format=lambda x: f"{x:g}"))

print("\ny normalized:")
print(y_norm_df.to_string(index=False, float_format=lambda x: f"{x:g}"))

print("\n=== GRADIENT DESCENT ===")
print("Bobot hasil Gradient Descent:")

for i, bobot in enumerate(w_gd, start=1):
    print(f"w{i} = {bobot:.10f}")

print(f"b  = {b_gd:.10f}")
print(f"Loss akhir     : {loss_gd[-1]:.3e}")
print(f"Jumlah iterasi : {len(loss_gd)}")
print(f"Prediksi data baru: {y_pred_gd[0]:.2f}%")
```

## Steepest Descent

```python
import numpy as np
import pandas as pd

# data x1, x2, x3
X_raw = np.array([
    [30, 9, 90],
    [20, 9, 80],
    [25, 3, 90]
], dtype=float)

# data target y
y_raw = np.array([92, 70, 85], dtype=float)

# data baru yang ingin diprediksi
X_new_raw = np.array([
    [27, 6, 85]
], dtype=float)

# normalisasi menggunakan min-max scaling
def minmax_fit(data):
  data_min = data.min(axis=0)
  data_max = data.max(axis=0)
  return data_min, data_max

def minmax_transform(data, data_min, data_max):
  return (data-data_min)/(data_max - data_min)

def minmax_inverse(data_norm, data_min, data_max):
  return data_norm*(data_max - data_min) + data_min

# normalisasi x
X_min, X_max = minmax_fit(X_raw)
X = minmax_transform(X_raw, X_min, X_max)

# normalisasi y
y_min, y_max = minmax_fit(y_raw)
y = minmax_transform(y_raw, y_min, y_max)

# normalisasi data baru
X_new = minmax_transform(X_new_raw, X_min, X_max)

# model
# y_hat = w1*x1 + w2*x2 + w3*x3 + b
def predict(X, w, b):
  return X @ w + b

# loss function
# f(w, b) = sum((y_hat - y)^2)
def loss_function(X, y, w, b):
  y_pred = predict(X, w, b)
  error = y_pred - y
  return np.sum(error ** 2)

# gradien
# grad = 2 * X.T @ (Xw - y)
def gradient(X, y, w, b):
  y_pred = predict(X, w, b)
  error = y_pred - y

  grad_w = 2 * X.T @ error
  grad_b = 2 * np.sum(error)

  return grad_w, grad_b

# steepest descent
def steepest_descent(X, y, max_iter=1000, tolerance=1e-10):
  jumlah_fitur = X.shape[1]
  w = np.zeros(jumlah_fitur)
  b = 0.0

  loss_history = []
  alpha_history = []

  for i in range(max_iter):
    loss = loss_function(X, y, w, b)
    loss_history.append(loss)

    grad_w, grad_b = gradient(X, y, w, b)

    direction_w = -grad_w
    direction_b = -grad_b

    y_pred = predict(X, w, b)
    error = y_pred - y

    j = X @ direction_w + direction_b

    alpha = - (j @ error) / (j @ j)
    alpha_history.append(alpha)

    w_new = w + alpha * direction_w
    b_new = b + alpha * direction_b

    if np.linalg.norm(w_new - w) < tolerance and abs(b_new - b) < tolerance:
      w = w_new
      b = b_new
      break

    w = w_new
    b = b_new

  return w, b, loss_history, alpha_history

# training model
w_sd, b_sd, loss_sd, alpha_sd = steepest_descent(X, y, max_iter=10000)

# prediksi
y_pred_sd_norm = predict(X_new, w_sd, b_sd)

# kembalikan ke skala persen
y_pred_sd = minmax_inverse(y_pred_sd_norm, y_min, y_max)

# output
print("=== DATA SETELAH NORMALISASI ===")

X_norm_df = pd.DataFrame(X, columns=["x1", "x2", "x3"])
y_norm_df = pd.DataFrame(y, columns=["y"])

print("\nX normalized:")
print(X_norm_df.to_string(index=False, float_format=lambda x: f"{x:g}"))

print("\ny normalized:")
print(y_norm_df.to_string(index=False, float_format=lambda x: f"{x:g}"))

print("\n=== STEEPEST DESCENT ===")
print("Bobot hasil Steepest Descent:")

for i, bobot in enumerate(w_sd, start=1):
    print(f"w{i} = {bobot:.10f}")

print(f"b  = {b_sd:.10f}")
print(f"Loss akhir     : {loss_sd[-1]:.3e}")
print(f"Jumlah iterasi : {len(loss_sd)}")

if len(alpha_sd) > 0:
    print(f"Alpha pertama  : {alpha_sd[0]:.10f}")

print(f"Prediksi data baru: {y_pred_sd[0]:.2f}%")
```
